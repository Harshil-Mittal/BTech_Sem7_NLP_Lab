```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Employe_Performance_dataset.csv')
```

```python
df.head()
```

```python
avg_df = df.groupby('Department', as_index=False)['Salary'].mean()
avg_df.columns = ['Department', 'Average Salary']
```

```python
sns.barplot(data=avg_df, x='Department', y='Average Salary')
plt.xticks(rotation=45)
plt.show()
```
